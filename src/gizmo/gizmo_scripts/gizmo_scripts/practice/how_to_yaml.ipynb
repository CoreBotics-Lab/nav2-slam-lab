```python
# Import the PyYAML library for reading and parsing YAML files
import yaml

# Absolute path to the saved waypoint YAML file
yaml_path = "/root/ros2_ws/src/gizmo/gizmo_navigation/maps/savedWaypointPose/test_waypoint.yaml"
```

```python
# Step 1: Open the YAML file in read mode ("r")
with open(yaml_path, "r") as file:
    # yaml.safe_load parses the file into standard Python objects (dict & list)
    data = yaml.safe_load(file)

# Expected YAML Structure:
# waypoints:
#   waypoint0:
#     pose: [2.396, 0.394, 0]
#     orientation: [0.937, 0, 0, 0.349] # [w, x, y, z]

# Step 2: Inspect types to see how YAML translates to Python:
print(type(data))                                           # Top-level is a dict
print(type(data["waypoints"]))                              # "waypoints" key maps to an inner dict
print(type(data["waypoints"].get("waypoint0")))             # "waypoint0" maps to an inner dict
print(type(data["waypoints"].get("waypoint0").get("pose"))) # "pose" maps to a list [x, y, z]
```

```python
# Extract the waypoints dictionary from top-level data
waypoints_dict = data["waypoints"]

# Checking key existence using the Pythonic "in" operator (O(1) hash lookup)
print("waypoint0" in waypoints_dict)   # True: "waypoint0" exists in keys
print("waypoint99" in waypoints_dict)  # False: "waypoint99" does not exist

# Iterating directly over a dictionary loops through its keys (waypoint names)
for wp_name in waypoints_dict:
    print(wp_name)
```

```python
# Iterate through (key, value) pairs using .items()
# Using .get() safely handles missing keys without throwing KeyError
for wp_name, wp_data in waypoints_dict.items():
    pose = wp_data.get("pose")                # List of position coordinates: [x, y, z]
    orientation = wp_data.get("orientation")  # List of quaternion values: [w, x, y, z]
    print(f"{wp_name}: pose={pose}, orientation={orientation}")
```
